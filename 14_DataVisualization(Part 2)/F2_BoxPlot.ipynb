```python
import matplotlib.pyplot as plt
import numpy as np
#BoxPLot
```

```python
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data)
plt.grid(True)
plt.title('Boxplot of Data')
plt.ylabel('Values')
plt.savefig('boxplot.png')
```

```python
#multiple datasets
group1 = np.random.normal(50,10,100)
group2 = np.random.normal(60,15,100)

plt.boxplot([group1,group2],tick_labels=["group1 ", "group2"],vert=False , showmeans=True)# by default vertical , but cdan change it to hirizontal
plt.show()
```

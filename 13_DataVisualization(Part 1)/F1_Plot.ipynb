```python
# For 2 diffrent datas 

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M
non_oscar_revenue  = [378,2788,185,275,400]
# Plot
plt.plot(years, oscar_revenue,"o--r",linewidth=3,label="Oscar revenue")# to plot #also format string 
                                           
plt.plot(years, non_oscar_revenue,"^-.g", label="non-Oscar revenue")# to plot  #also formnat string 
plt.title("Oscar Movies in the each Year") #for title 

plt.xlabel("Years")
plt.ylabel("Revenue id( in m$)")

plt.legend(loc="upper right")
plt.show()
```

```python
#learning tight_layout()
# Oscar Winner Movies
movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232] # in $M

plt.plot(movies,revenue )
plt.title("Movie revenue")
plt.xlabel("Years")
plt.ylabel("Revenue id( in m$)")
plt.tight_layout()
plt.savefig("movies_plot.png") #saves the figure ploted
plt.show()
    
```

```python
#Pie chart
import matplotlib.pyplot as plt
```

```python
#https://matplotlib.org/stable/api/_as_gen/matplotlib.patches.Wedge.html 
expenses=["salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amount = [500,150,120,100,50]

colors =["blue", "pink", "red", "yellow", "green"]
explode=[0,0,0,0.2,0] #it will highlight the slice
plt.style.use("default")
plt.pie(amount ,labels=expenses ,autopct="%1.1f%%" ,
       wedgeprops ={ 
           "edgecolor":"black",
           "linewidth": 1,
           "linestyle": "-.",
       },
        explode=explode, #it will highlight the slice by which u have mentioned above
        startangle=90, #moved to angle 90
        shadow = True #gives 3d effect
       )
plt.style.available # to see which styles are available 
plt.show()
```

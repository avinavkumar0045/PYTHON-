#BAR CHART

```python
import matplotlib.pyplot as plt
```

```python
#BAR chart
movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232] # in $M

plt.bar(years, revenue,color="orange")

plt.title("Movie revenue")
plt.xlabel("Year")
plt.ylabel("Revenue id( in m$)")
for i in range (len(years)) : 
     plt.text(years[i],revenue[i]+10,str(revenue[i]),ha="center") #plt.text()
plt.ylim(0 ,revenue[0]+100) #to define the limits of the the graph, # can be same for x 
plt.xlim(min(years)-1 , max(years)+1)

```

```python
#BAR chart
import numpy as np

movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M
non_oscar_revenue  = [378,2788,185,275,400]

x = np.arange(len(years)) #to get the range for the numpy problem
width = 0.4 #phele width ko small kro

plt.bar(x-width/2 , oscar_revenue,width,label="Oscar movies") #x-(width/2) = 0.5 - 0.2 = 0.3
plt.bar(x+width/2 , non_oscar_revenue , width,label="Non -Oscar movies")# 0.5 + 0.2 = 0.7

plt.title("Oscar vs non oscar movies")
plt.xlabel("year")
plt.ylabel("Revenue(in m$)")
plt.legend()
plt.xticks(x,years) # lables ke jagah orginal data ko replace krta hai.
```

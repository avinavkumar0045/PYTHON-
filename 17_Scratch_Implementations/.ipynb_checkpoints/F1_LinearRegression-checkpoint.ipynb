# Linear Regression with Gradient Descent

```python
import numpy as np
```

```python
#Linear Regression
class LinearRegression:
    def __init__(self , learning_rate = 0.01 , n_iter =1000): #learning_rate= alpha
        self.bias = None
        self.weights  = None
        self.lr = learning_rate
        self.n_iter = n_iter

    def fit(self , X ,y):
        m,n = X.shape#(num of samples , no of features)

        #STEP 1- initialize parameters
        self.bias =0
        self.weights = np.zeros(n)

        #gradient descent
        for i in range(self.n_iter): #as step 2,3,4 should be repeat4ed n_iter times
            #step 2 calc y_pred
            y_pred = self.bias + np.dot(X , self.weights); #bias + dot product

            #step 3 - calc gradients
            db = (1/m)* np.sum(y_pred - y)
            dw = (1/m)* np.dot(X.T , (y_pred - y)) # we do the transpose of the matrix to get the dot product 
            

            #step 4 - convergence theorem applied and parameters are updated
            self.bias -= self.lr * db #.lr is learning rate(alpha)
            self.weights -= self.lr * dw


    def predict(self , X):
        y_pred =  self.bias+ np.dot(X, self.weights)
        return y_pred
```

```python
#applying sameple data
X = np.array([[1],[2],[3],[4],[5]])
y = np.array([2,4,6,8,10])

model = LinearRegression()
model.fit(X ,y)

y_pred = model.predict(X)
print(y_pred)

print(model.bias)
print(model.weights)
```

# Linear Regression with 'Ordinary Least Squares (OLS)'

```python
class LinearRegressionOLS:
    def __init__(self):
        self.bias = None;
        self.weights = None;

    def fit(self , X,y):
        m,n = X.shape

        X_b = np.c_[np.ones((m,1)), X] # updated X
        # normal equation
        theta = np.dot( np.linalg.inv (np.dot(X_b.T,X_b)) , np.dot(X_b.T , y)) 
        # theta =  np.linealg.inv(X_b.T @X) @ X_b.T @ y # latest way to find dot product in python use "@"

        self.bias  = theta[0]
        self.weights = theta[1:]
        
        

    def predict(self , X):
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred
```

```python
#applying sameple data
X = np.array([[1],[2],[3],[4],[5]])
y = np.array([2,4,6,8,10])

model = LinearRegressionOLS()
model.fit(X ,y)

y_pred = model.predict(X)
print(y_pred)

print(model.bias)
print(model.weights)
```

# DBSCAN clustering

```python
from sklearn.datasets import load_iris
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()
X = iris.data
y = iris.target
```

```python
# scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# visualize
sns.scatterplot(x = X_scaled[:,0] , y=X_scaled[:,2], c= y)
```

```python
# DBSCAN 
# in this iris dataset as we know the ans it does not perform well ,but performs very well in non-linear datasets
from sklearn.cluster import DBSCAN
dbscan  =DBSCAN(
    eps = 0.8,
    min_samples =5,
)
```

```python
labels = dbscan.fit_predict(X_scaled)
sns.scatterplot(x = X_scaled[:,0] , y=X_scaled[:,2], c= labels)
```

 # DBSCAN on Non-linear data 

```python
from sklearn.datasets import make_moons
```

```python
X , y = make_moons(
    n_samples= 300,
    noise = 0.05,
    random_state = 42
)
```

```python
X_scaled = scaler.fit_transform(X)
sns.scatterplot(x = X_scaled[:,0] , y = X[:,1])
```

```python
# k - Means
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(
    n_clusters =2,
    random_state = 42
)
labels = kmeans.fit_predict(X_scaled)
```

```python
sns.scatterplot(x = X_scaled[:,0], y = X_scaled[:,1], c =labels) 
```

```python
# as we can see that it somewhere divides the data in to linear format , let see woth DBSCAN what happens
```

```python
# DBSCAN 
from sklearn.cluster import DBSCAN
dbscan  =DBSCAN(
    eps = 0.8,
    min_samples =5,
)
label = dbscan.fit_predict(X_scaled)
```

```python
sns.scatterplot(x = X_scaled[:,0], y = X_scaled[:,1], c =label) 
```

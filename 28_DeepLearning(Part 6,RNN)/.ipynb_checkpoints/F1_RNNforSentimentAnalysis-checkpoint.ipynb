# Pre-processing of data

## 1.Converting to lowercase

```python
df["review"] = df["review"].str.lower()
```

## 2.Removing URL's

```python
import re # regular expression

#sample text = 'abc is the word , abc" , # abc => zyz
# new_text = re.sub("abc" , "xyz" , sample_text)
```

```python
def remove_urls(text):
    text = re.sub( r"http\S+" ,"", text) # patter, replacement , string)
    return text

df["review"]= df["review"].apply(remove_urls)
```

## 3.Remove Punctuations

```python
def remove_punctuations(text):
    text = re.sub( r"[^A-Za-z0-9\s]" ,"", text) # A-Z ,a-z,0-9 will not be there when there is puctuations 
    return text
    
df["review"]= df["review"].apply(remove_punctuations)
```

## 4.Remove html tags

```python
def remove_html(text): 
    text = re.sub( r"<.*?>" ,"", text) # A-Z ,a-z,0-9 will not be there when there is puctuations 
    return text
    
df["review"]= df["review"].apply(remove_html)
df.head()
```

## 5. Removing the stopwords

```python
import nltk # natural language tokenizer
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

```

```python
from nltk.tokenize import word_tokenize # tokenizes the word
from nltk.corpus import stopwords
# importds the work
df.head()
```

```python
# sample_text="I like coding in Java"
# tokens = word_tokenize(sample_text)
# tokens


def remove_stopwords(text):
    tokens = word_tokenize(text)
    stop_words = stopwords.words("english")

    for word in tokens:
        if word in stop_words:
            text = text.replace(word, "")
    return text

df["review"] = df["review"].apply(remove_stopwords)

```

## 6. Performing Stemming

```python
# running -> run , played->play , 

#PorterStemming

from nltk.stem import PorterStemmer
```

```python
def stemming(text):
    ps = PorterStemmer()
    stemmed_words= []

    tokens = word_tokenize(text)
    for token in tokens:
        stemmed_token = ps.stem(text)
        stemmed_words.append(stemmed_token)
        
    return  " ".join(stemmed_words)

df["review"] = df["review"].apply(stemming)
    
```
